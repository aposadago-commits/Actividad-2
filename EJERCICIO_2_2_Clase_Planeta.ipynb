EJERCICIO 2.2 - Clase Planeta

```python
from enum import Enum

class TipoPlaneta(Enum):
    GASEOSO = "GASEOSO"
    TERRESTRE = "TERRESTRE"
    ENANO = "ENANO"


class Planeta:
    """Planeta del sistema solar. Distancia al Sol en millones de km."""

    UA_MILLONES_KM = 149.597870          # 1 UA = 149 597 870 km
    LIMITE_CINTURON_UA = 3.4             # el cinturón de asteroides llega hasta 3.4 UA

    def __init__(self, nombre=None, cantidad_satelites=0, masa=0.0, volumen=0.0,
                 diametro=0, distancia_sol=0, tipo=None, es_observable=False):
        self.nombre = nombre
        self.cantidad_satelites = cantidad_satelites
        self.masa = masa                    # kg
        self.volumen = volumen              # km^3
        self.diametro = diametro            # km
        self.distancia_sol = distancia_sol  # millones de km
        self.tipo = tipo
        self.es_observable = es_observable

    def imprimir(self):
        print(f"Nombre del planeta = {self.nombre}")
        print(f"Cantidad de satélites = {self.cantidad_satelites}")
        print(f"Masa del planeta = {self.masa}")
        print(f"Volumen del planeta = {self.volumen}")
        print(f"Diámetro del planeta = {self.diametro}")
        print(f"Distancia al sol = {self.distancia_sol}")
        print(f"Tipo de planeta = {self.tipo.value if self.tipo else None}")
        print(f"Es observable = {self.es_observable}")

    def calcular_densidad(self):
        return self.masa / self.volumen

    def es_planeta_exterior(self):
        """Exterior = más allá del cinturón de asteroides (> 3.4 UA)."""
        limite = self.LIMITE_CINTURON_UA * self.UA_MILLONES_KM
        return self.distancia_sol > limite


def ejercicio_2_2():
    print("=" * 60)
    print("EJERCICIO 2.2 - Planeta")
    print("=" * 60)
    tierra = Planeta("Tierra", 1, 5.9736e24, 1.08321e12, 12742, 150,
                     TipoPlaneta.TERRESTRE, True)
    tierra.imprimir()
    print(f"Densidad del planeta = {tierra.calcular_densidad()}")
    print(f"Es planeta exterior = {tierra.es_planeta_exterior()}")
    print()

    jupiter = Planeta("Júpiter", 79, 1.899e27, 1.4313e15, 139820, 750,
                      TipoPlaneta.GASEOSO, True)
    jupiter.imprimir()
    print(f"Densidad del planeta = {jupiter.calcular_densidad()}")
    print(f"Es planeta exterior = {jupiter.es_planeta_exterior()}")
    print()

ejercicio_2_2()
```
